```python
import cv2
import numpy as np
# from colab.patches import cv2_imshow # Removed this import
from google.colab.patches import cv2_imshow
# Load image
image = cv2.imread('download.jpg')  # Replace with your image path
if image is None:
    raise FileNotFoundError("Image not found or invalid path")

# 1. Display original image
cv2_imshow( image) # Changed to cv2.imshow

# 2. Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow( gray)

# 3. Resize image (to 300x300)
resized = cv2.resize(image, (300, 300))
cv2_imshow( resized)

# 4. Crop image (first 200 rows and columns)
cropped = image[0:200, 0:200]
cv2_imshow( cropped)

# 5. Rotate image by 45 degrees
h, w = image.shape[:2]
M = cv2.getRotationMatrix2D((w//2, h//2), 45, 1.0)
rotated = cv2.warpAffine(image, M, (w, h))
cv2_imshow( rotated)

# 6. Flip image horizontally
flipped = cv2.flip(image, 1)
cv2_imshow( flipped)

# 7. Thresholding (convert to binary)
_, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)
cv2_imshow( binary)

# 8. Histogram equalization
equalized = cv2.equalizeHist(gray)
cv2_imshow( equalized)

# 9. Gaussian blur (5x5 kernel)
blurred = cv2.GaussianBlur(image, (5, 5), 0)
cv2_imshow( blurred)

# 10. Edge detection (Canny)
edges = cv2.Canny(gray, 100, 200)
cv2_imshow( edges)

# 11. Dilation (3x3 kernel)
kernel = np.ones((3, 3), np.uint8)
dilated = cv2.dilate(binary, kernel)
cv2_imshow( dilated)

# 12. Erosion (3x3 kernel)
eroded = cv2.erode(binary, kernel)
cv2_imshow( eroded)

# 13. Convert to HSV color space
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
cv2_imshow( hsv)

# 14. Add random noise
noise = np.random.normal(0, 25, image.shape).astype(np.uint8)
noisy_img = cv2.add(image, noise)
cv2_imshow( noisy_img)

# 15. Adjust brightness (+200)
brightened = cv2.convertScaleAbs(image, alpha=1, beta=50)
cv2_imshow( brightened)


```
